# Byte Pair Encording 


- Nature language processing aims to make machines understand text at human level. 
- Machines only understand numbers not texts. 
- **Tokenization** is process of converting text into numbers. 

- There are several types of tokenization are there, they are: 
    1. Character tokenization 
    2. Word tokenization 
    3. Sub word tokenization 
        - Byte pair tokenization 
        - WordPiece tokenization 

### 1. Character Tokenization

Character tokenization breaks down text into individual characters.Each character in the text becomes a token.
- Example: "Hello" → ["H", "e", "l", "l", "o"]

*Pros:*
- Handles OOV words well: Any word can be represented as a sequence of characters.
- Small vocabulary: The vocabulary size is fixed and small.
- Good for morphologically rich languages: Can capture nuances within words.

*Cons:*

- Loss of semantic meaning: Individual characters carry very little semantic information, making it harder for models to learn meaningful representations.
- Longer sequences: Sentences become much longer sequences of tokens, increasing computational cost and complexity for models.


### 2. Word Tokenization

Word tokenization is the most straightforward and common type of tokenization. It involves splitting text into individual words based on spaces and punctuation.
Tokenizer splits the text whenever it encounters a whitespace character. Punctuation marks are often treated as separate tokens or are removed.

- Example: "Hello, world!" → ["Hello", ",", "world", "!"]

*Pros:*

- Simple to implement and understand.

*Cons:*

- Vocabulary size: Leads to a very large vocabulary, especially for languages with rich morphology (e.g., German, Turkish), or when dealing with misspellings or rare words.
- Out-of-vocabulary (OOV) words: Struggles with words not seen during training, leading to unknown tokens.
- Morphological variations: Different forms of the same word (e.g., "run," "running," "ran") are treated as distinct tokens, increasing vocabulary size and potentially losing semantic relationships.

### 3. Subword Tokenization

Aims to strike a balance between word and character tokenization. It breaks down words into smaller, frequently occurring subword units. This approach is particularly effective in handling OOV words and reducing vocabulary size while retaining more semantic information than character tokenization.

1. Byte Pair Encoding (BPE)

    - It's merges the most frequent pairs of characters or character sequences into new subword units. 
    - It starts with a vocabulary of individual characters and repeatedly merges the most frequent adjacent pair of tokens in the training data until a predefined vocabulary size is reached or no more merges are possible.

    - Example: If "lo" and "wer" are frequent, they might merge to "lower". If "low" and "est" are frequent, they might become "lowest".

*Pros:*

- Handles OOV words: Unknown words are broken down into known subwords or individual characters.

- Manages vocabulary size: Creates a manageable vocabulary that captures common word parts.

- Good for morphology: Can learn common prefixes, suffixes, and root words.

*Cons:*
- Can sometimes create arbitrary splits that don't align with linguistic intuition.


*Used in: GPT-2, RoBERTa*
