### Note:
The tokenizer is completely seperate, independent module from LLM. It has  its own training dataset of text(which   could be different from that of LLM), on which you train the vocabulary using Byte Pair encording (BPE) algorithm. It then translate back and forth between raw text and sequence of tokens. The LLM later only ever sees   the tokens and never directly deals with any text. 


- Tokenizer is translation layer between raw text (which is sequence of unicode points). It can take raw text and translate into sequence of tokens and vice versa. 
- After training the tokenizer we can do encoding and decoding. 

1. The converted sequence of tokens those who stored into disk, and thats what LLM is actually reading when its training on them. 

#### Many modern tokenizers, especially those used in LLMs, are based on algorithms like Byte-Pair Encoding (BPE). In BPE, the training process iteratively merges the most frequent adjacent character or token pairs into new, single tokens.

1. #### Number of Merges and Language Representation:
    - The number of merges performed during BPE training can indirectly influence how many languages are effectively covered and to what depth.
    - A larger vocabulary size (often a result of more merges) allows for more distinct tokens, which can better represent the intricacies of numerous languages and coding constructs.

2. #### Merges and Token Density: The merges determine the density with which specific data types (like a particular language or code) are represented in the token space.

    - Example: If the training dataset includes a significant amount of Japanese text, the BPE algorithm will identify and merge frequent Japanese character sequences into specific subword tokens. ***This results in shorter sequences of tokens for Japanese text compared to a tokenizer not trained sufficiently on Japanese.***

    - Shorter token sequences mean more efficient processing and often better model performance for that language. Conversely, if a language is under-represented, its tokens might be broken down into many smaller, less meaningful units, leading to longer sequences and potentially hindering model performance.

```python
vocab = {idx:bytes([idx]) for idx in range(256)}
for (chr1,chr2),idx in merged_token.items():
    vocab[idx] = vocab[chr1] + vocab[chr2]
print(vocab) 
```

```python
def decode(ids:list[int]) -> str:
    """Given ids (list of integers), return python string"""
    tokens  = b"".join(vocab[idx] for idx in ids)    # b"".join ==> concatenate bytes together 
    text    = tokens.decode("utf-8",errors="replace")
    return text
```

```python
def encode(text:str)->list[int]:
    """Given a strig, return list of integers (tokens)"""
    tokens  = list(text.encode("utf-8"))
    while len(text)>=2:
        stats   = get_stats(tokens)
        pair    = min(stats,key=lambda pair: merged_token.get(pair,float("inf")))   # if no value present, it return the  first value from 'stats'
        if pair not in merged_token:
            break   # nothing else can be merged 
        idx     = merged_token[pair]
        tokens  = merge(tokens,pair,idx)
    
    return tokens    
```

## Note: 
*The decoded and then re-encoded version of a string might not be the same as the original string, especially if the original string contained characters that were not decodable by the specified encoding. Not all token sequence are valid **UTF-8** bytes of streams, So some of them can't be decodable.*

### The Difference

### Here's why decode() -> encode() might result in a different string:

- Undecodable Characters and Error Handling: If your initial decoding process encountered undecodable characters and you used an error handling strategy like 'ignore' or 'replace', the resulting string will be different from the original byte sequence's intended meaning.

    - If you ignored undecodable bytes, those characters are gone from the decoded string. When you re-encode this string, those characters won't magically reappear.

    - If you replaced undecodable bytes with a replacement character (like ``), then the re-encoded string will contain the byte representation of this replacement character, not the original undecodable byte sequence.
