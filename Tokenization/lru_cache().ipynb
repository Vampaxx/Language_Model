## lru_cache()

```python
from functools import lru_cache

@lru_cache(maxsize=128)  # cache up to 128 unique calls
def fibonacci(n):
    print(f"Computing fib({n})")  # to show when it's actually computed
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(10))
```

- 🧠 What happens here?

- When fibonacci(10) is called:

    - It internally calls fibonacci(9), fibonacci(8), etc.

- Once a value like fibonacci(5) is computed, it's cached.
- If fibonacci(5) is needed again, Python just returns the cached result — no recomputation.

```python
import time 
from functools import lru_cache
```

```python
def slow_square(x):
    time.sleep(1)
    return x**x 

@lru_cache()
def fast_square (x):
    time.sleep(1)
    return x**x 
```

```python
start = time.time()
print(slow_square(12))
print(slow_square(12))
print(f"without caching {time.time() - start}")

start = time.time()
print(fast_square(12))
print(fast_square(12))
print(f"with caching {time.time() - start}")
```
